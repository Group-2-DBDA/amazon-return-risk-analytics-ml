1️⃣ Dataset Overview & Data Quality

Columns used

All top-level columns

What to analyze

- Total number of products

- Unique parent_asin

- Missing value percentage per column

- Identify flat vs nested vs array columns

```python
meta_df.printSchema()
meta_df.select("details").show(3, truncate=False)
meta_df.selectExpr("details.*").show(5, truncate=False)
```

```python
meta_row_count = meta_df.count()
meta_col_count = len(meta_df.columns)
print(f"Total rows: {meta_row_count}, Total columns: {meta_col_count}")
meta_df.count()
```

```python
# Total number of products
total_products = meta_df.count()

# Unique parent_asin
unique_parent_asin = meta_df.select("parent_asin").distinct().count()

# For display, create a summary DataFrame
import pandas as pd

missing_data = []
for col in meta_df.columns:
    non_null_count = meta_df.filter(f"{col} IS NOT NULL").count()
    percent_missing = 100 * (1 - (non_null_count / total_products))
    missing_data.append((col, percent_missing))
missing_df = pd.DataFrame(missing_data, columns=["column", "missing_percent"])
display(missing_df)

# Identify flat vs nested vs array columns
schema = meta_df.schema
flat_cols = []
nested_cols = []
array_cols = []

for field in schema.fields:
    if str(field.dataType).startswith("StructType"):
        nested_cols.append(field.name)
    elif str(field.dataType).startswith("ArrayType"):
        array_cols.append(field.name)
    else:
        flat_cols.append(field.name)

overview_df = pd.DataFrame({
    "flat_columns": [flat_cols],
    "nested_columns": [nested_cols],
    "array_columns": [array_cols]
})
display(overview_df)

# Display summary stats
print(f"Total products: {total_products}")
print(f"Unique parent_asin: {unique_parent_asin}")
```

2️⃣ Category & Taxonomy Analysis
Columns used

- main_category

- categories (array)

What to analyze

- Distribution of products by main_category

- Most frequent category paths

- Depth of category hierarchy

```python
from pyspark.sql.functions import col, size, expr, explode, concat_ws

# Distribution of products by main_category
main_cat_dist = meta_df.groupBy("main_category").count().orderBy(col("count").desc())
display(main_cat_dist)

# Most frequent category paths (categories is an array of strings)
category_paths = meta_df.withColumn("category_path", concat_ws(" > ", col("categories"))) \
.groupBy("category_path").count().orderBy(col("count").desc())
display(category_paths)

# Depth of category hierarchy (length of categories array)
category_depth = meta_df.withColumn("category_depth", size(col("categories"))) \
                        .groupBy("category_depth").count().orderBy(col("category_depth"))
display(category_depth)
```

3️⃣ Pricing Analysis

Columns used

- price

What to analyze

- Price distribution

- Min / max / median price

- % products with missing price

- Outliers (extremely high/low prices)

```python
from pyspark.sql.functions import col, count, isnan, when, min as spark_min, max as spark_max, percentile_approx

# Price distribution
display(meta_df.select("price"))

# Min / max / median price
price_stats = meta_df.select(
    spark_min(col("price")).alias("min_price"),
    spark_max(col("price")).alias("max_price"),
    percentile_approx(col("price"), 0.5).alias("median_price")
)
display(price_stats)

# % products with missing price
total_products = meta_df.count()
missing_price_count = meta_df.filter(col("price").isNull() | isnan(col("price"))).count()
missing_price_percent = 100 * missing_price_count / total_products
display(spark.createDataFrame([(missing_price_count, missing_price_percent)], ["missing_price_count", "missing_price_percent"]))

# Outliers (extremely high/low prices)
# Define outliers as outside 1.5*IQR
quantiles = meta_df.approxQuantile("price", [0.25, 0.75], 0.01)
q1, q3 = quantiles
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers_df = meta_df.filter((col("price") < lower_bound) | (col("price") > upper_bound))
display(outliers_df.select("price"))
```

```python
import matplotlib.pyplot as plt

# Collect price data for histogram (excluding missing/null values)
price_values = meta_df.filter(col("price").isNotNull() & ~isnan(col("price"))).select("price").toPandas()["price"]

plt.figure(figsize=(8, 5))
plt.hist(price_values, bins=50, color='skyblue', edgecolor='black')
plt.xlabel('Price')
plt.ylabel('Number of Products')
plt.title('Price Distribution Histogram')
plt.grid(True)
plt.show()

# Display missing price percentage
total_products = meta_df.count()
missing_price_count = meta_df.filter(col("price").isNull() | isnan(col("price"))).count()
missing_price_percent = 100 * missing_price_count / total_products

print(f"Missing price %: {missing_price_percent:.2f}%")
```

4️⃣ Product Popularity & Reliability Analysis

Columns used

- average_rating

- rating_number

What to analyze

- Distribution of average_rating

- Distribution of rating_number

- Rating vs popularity relationship

Why this matters

Distinguishes trusted vs risky products.

```python
from pyspark.sql.functions import col, isnan, percentile_approx, when

# --------------------------------------------------
# 1. Clean Data: remove NULLs and NaNs
# --------------------------------------------------

clean_df = meta_df.select("average_rating", "rating_number") \
    .filter(
        col("average_rating").isNotNull() &
        col("rating_number").isNotNull() &
        ~isnan(col("average_rating")) &
        ~isnan(col("rating_number"))
    )

# --------------------------------------------------
# 2. Distribution of average_rating
# --------------------------------------------------

display(clean_df.select("average_rating"))

# Optional: bucket average_rating for clearer distribution
avg_rating_bucket_df = clean_df.withColumn(
    "average_rating_bucket",
    when(col("average_rating") < 2, "1–2")
    .when(col("average_rating") < 3, "2–3")
    .when(col("average_rating") < 4, "3–4")
    .otherwise("4–5")
)

display(
    avg_rating_bucket_df.groupBy("average_rating_bucket")
    .count()
    .orderBy("average_rating_bucket")
)

# --------------------------------------------------
# 3. Distribution of rating_number
# --------------------------------------------------

display(clean_df.select("rating_number"))

# Optional: popularity buckets
rating_number_bucket_df = clean_df.withColumn(
    "rating_number_bucket",
    when(col("rating_number") < 10, "<10")
    .when(col("rating_number") < 100, "10–99")
    .when(col("rating_number") < 1000, "100–999")
    .otherwise("1000+")
)

display(
    rating_number_bucket_df.groupBy("rating_number_bucket")
    .count()
    .orderBy("rating_number_bucket")
)

# --------------------------------------------------
# 4. Median Statistics
# --------------------------------------------------

median_stats_df = clean_df.select(
    percentile_approx(col("average_rating"), 0.5).alias("median_average_rating"),
    percentile_approx(col("rating_number"), 0.5).alias("median_rating_number")
)

display(median_stats_df)

# --------------------------------------------------
# 5. Min / Max Statistics
# --------------------------------------------------

min_max_df = clean_df.select(
    percentile_approx("average_rating", 0.0).alias("min_average_rating"),
    percentile_approx("average_rating", 1.0).alias("max_average_rating"),
    percentile_approx("rating_number", 0.0).alias("min_rating_number"),
    percentile_approx("rating_number", 1.0).alias("max_rating_number")
)

display(min_max_df)

# --------------------------------------------------
# 6. Rating vs Popularity Relationship
# --------------------------------------------------

rating_popularity_df = clean_df.select(
    "average_rating",
    "rating_number"
)

display(rating_popularity_df)
```

```python
import matplotlib.pyplot as plt

# Collect data for scatter plot: average_rating vs rating_number
scatter_df = clean_df.select("average_rating", "rating_number").toPandas()
# Distribution of price
price_values = meta_df.filter(col("price").isNotNull() & ~isnan(col("price"))).select("price").toPandas()["price"]
plt.figure(figsize=(8, 5))
plt.hist(price_values, bins=50, color='skyblue', edgecolor='black')
plt.xlabel('Price')
plt.ylabel('Number of Products')
plt.title('Price Distribution Histogram')
plt.grid(True)
plt.show()

# Distribution of average_rating
avg_rating_values = clean_df.select("average_rating").toPandas()["average_rating"]
plt.figure(figsize=(8, 5))
plt.hist(avg_rating_values, bins=20, color='orange', edgecolor='black')
plt.xlabel('Average Rating')
plt.ylabel('Number of Products')
plt.title('Average Rating Distribution')
plt.grid(True)
plt.show()

# Distribution of rating_number
rating_number_values = clean_df.select("rating_number").toPandas()["rating_number"]
plt.figure(figsize=(8, 5))
plt.hist(rating_number_values, bins=50, color='green', edgecolor='black')
plt.xlabel('Rating Number')
plt.ylabel('Number of Products')
plt.title('Rating Number Distribution')
plt.grid(True)
plt.show()
plt.figure(figsize=(8, 5))
plt.scatter(scatter_df["average_rating"], scatter_df["rating_number"], alpha=0.5, color='teal')
plt.xlabel("Average Rating")
plt.ylabel("Rating Number")
plt.title("Average Rating vs Rating Number")
plt.grid(True)
plt.show()
```
